```python
def get_grade(score): 
    if score >= 80: 
        return "Distinction" 
    elif score >= 70: 
        return "Merit" 
    elif score >= 60: 
        return "Pass" 
    else: 
        return "Fail" 
```

```python
print(get_grade(85)) 
print(get_grade(72)) 
print(get_grade(45))
```

```python
import numpy as np 
arr = np.array([10, 20, 30, 40, 50])
```

```python
import pandas as pd 
data = { 
"StudentID": [101, 102, 103, 104, 105, 106], 
"Name": ["Alice", "Bob", "Charlie", "Diana", "Eve", "Frank"], 
"Age": [25, 30, 35, 28, 32, 27], 
"Score": [85, 72, 90, 68, 95, 78], 
"Department": ["CS", "IT", "CS", "Business", "IT", "CS"] 
} 
df = pd.DataFrame(data) 
```

```python
print(df.head()) 
print(df.shape) 
print(df.columns) 
print(df.dtypes) 
```

Display only the Name and Score columns 

```python
print(df[["Name", "Score"]])
    
```

```python
Display the first three rows using iloc. 
```

```python
print(df.iloc[:3])
```

Filter students with Score greater than 80. 

```python
print(df[df["Score"]>80])
```

Filter students whose Department is CS. 

```python
print(df[df["Department"] == "CS"])
```

Sort the DataFrame by Score in descending order. 

```python
sorted_df = df.sort_values(by="Score", ascending=False)
print(sorted_df)
```
